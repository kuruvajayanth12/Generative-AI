```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame([{"Name": "Jayanth", "Age": 21, "city": "Bangalore"},{"Name": "Sumo", "Age": 30, "city": "Bangalore"},{"Name": "Sai", "Age": 26, "city": "Bangalore"}])
plt.figure(figsize=(5,3))
plt.bar(data["Name"], data["Age"])
plt.title("All Ages")
plt.xlabel("Name")
plt.ylabel("Age")
plt.show()
above_25 = data[data["Age"] > 25]
plt.figure(figsize=(5,3))
plt.bar(above_25["Name"], above_25["Age"])
plt.title("People with Age Above 25")
plt.xlabel("Name")
plt.ylabel("Age")
plt.show()
```
